```python
import pandas as pd
# Imports pandas, which we use to read, filter, and combine the CSV files.

import os
# Imports os, which we use to work with the filenames.
```

```python
def load_christchurch(file_name):
    # Defines a function that loads and prepares one CSV dataset.
    # file_name is the name of the CSV file being processed.

    df = pd.read_csv(file_name)
    # Reads the CSV file into a pandas DataFrame called df.

    df = df[df["neighbourhood_group"] == "Christchurch City"].copy()
    # Filters the dataset to keep only listings from Christchurch City.
    # .copy() creates an independent copy of the filtered data.

    month_year = os.path.basename(file_name).replace("_listings.csv", "").upper()
    # Gets the filename without the folder path.
    # Removes "_listings.csv" from the filename.
    # Converts the remaining month/year text to uppercase.
    # For example, "Oct26_listings.csv" becomes "OCT26".

    df["month_year"] = month_year
    # Creates a new column called month_year.
    # Every row from this dataset receives its corresponding month/year.

    return df
    # Returns the processed Christchurch dataset.
```

```python
files = [
    "Oct26_listings.csv",
    "Nov26_listings.csv",
    "Dec26_listings.csv",
    "Jan26_listings.csv",
    "Feb26_listings.csv",
    "Mar26_listings.csv",
    "April26_listings.csv",
    "May26_listings.csv",
    "June26_listings.csv"
]
# Stores the names of all nine monthly CSV files.
# These are the datasets covering October 2025 to June 2026.
```

```python
christchurch_datasets = []
# Creates an empty list.
# Each processed Christchurch dataset will be stored in this list.

for file in files:
    # Loops through each CSV filename in the files list.

    df = load_christchurch(file)
    # Sends the current filename to our function.
    # The function loads the file, filters Christchurch,
    # and adds the month_year column.

    christchurch_datasets.append(df)
    # Adds the processed DataFrame to the list.
```

```python
christchurch_all = pd.concat(christchurch_datasets, ignore_index=True)
# Combines all nine processed Christchurch datasets into one larger DataFrame.
# ignore_index=True creates a new continuous row index for the combined dataset.
```

```python
print(christchurch_all.shape)
# Displays the total number of rows and columns in the combined dataset.

print(christchurch_all["month_year"].unique())
# Displays all the month/year values present in the combined dataset.

print(christchurch_all["month_year"].value_counts())
# Counts how many Christchurch listings came from each monthly dataset.
# This helps verify that all nine datasets were successfully included.
```

```python
christchurch_all.to_csv(
    "christchurch_listings_oct2025_june2026.csv",
    index=False
)
# Saves the final combined DataFrame as a CSV file.
#
# "christchurch_listings_oct2025_june2026.csv" is the name
# of the new combined CSV file.
#
# index=False prevents pandas from saving its row index
# as an additional column in the CSV.
```
